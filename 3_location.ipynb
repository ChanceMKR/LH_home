```python
jinju = pd.read_csv('./data/jinju.csv', index_col = 0)
jinju = jinju.iloc[:4]
print(jinju.shape)
jinju
```

```python
# 안전하게 osrm_distance 호출하는 래퍼 함수
def safe_osrm_distance(origin, dest):
    try:
        return osrm_distance(origin, dest)
    except Exception as e:
        # 에러 메시지 출력(필요시 삭제)
        print(f"⚠️ 거리 계산 오류: {origin} → {dest} ({e})")
        return None

# 반복문을 돌며 각 sent 주소마다 컬럼 생성
for idx, sent_row in sent.reset_index(drop=True).iterrows():
    keyword = sent_row.keyword.replace(" ", "_")
    col_name = f"dist_{keyword}"
    addr = sent_row['address']
    
    # apply 대신 progress_apply 써도 됩니다
    jinju[col_name] = jinju['location'].apply(
        lambda rn: safe_osrm_distance(rn, addr)
    )

# 결과 확인
jinju
```
